### 4.1 Investigasi Missing Values

Sebelum melakukan pengisian atau penghapusan data, diperlukan pemeriksaan untuk menentukan apakah missing values menunjukkan kesalahan pencatatan atau kondisi bisnis yang valid.

Investigasi awal difokuskan pada `voucher_type` dan `referral_code_used` karena kedua kolom tersebut relevan terhadap analisis promosi dan akuisisi pengguna.

```python
# CHECK MISSING VALUE PATTERNS

print("=== VOUCHER TYPE VS PROMO AMOUNT ===")

display(
    transactions.groupby(
        transactions["voucher_type"].isna()
    )["promo_amount"].agg(
        ["count", "mean", "median", "min", "max"]
    ).rename(index={
        False: "Voucher recorded",
        True: "Voucher missing"
    })
)

print("\n=== REFERRAL CODE VS ACQUISITION CHANNEL ===")

display(
    pd.crosstab(
        users["acquisition_channel"],
        users["referral_code_used"].isna(),
        normalize="index"
    ).mul(100).round(2).rename(
        columns={
            False: "Referral code recorded (%)",
            True: "Referral code missing (%)"
        }
    )
)
```

### 4.2 Hasil Investigasi Missing Values

Hasil pemeriksaan menunjukkan bahwa seluruh **113.288 transaksi (41,97%)** dengan `voucher_type` kosong memiliki nilai `promo_amount` sebesar Rp0. Hal ini menunjukkan bahwa kekosongan tersebut konsisten dengan transaksi tanpa penggunaan voucher promosi.

Sementara itu, analisis `acquisition_channel` menunjukkan bahwa seluruh pengguna dari kanal `referral` mempunyai kode referral tercatat, sedangkan pengguna dari kanal `ads`, `affiliate`, dan `organic` tidak mempunyai kode referral.

Berdasarkan pola tersebut, missing values pada `voucher_type` dan `referral_code_used` dapat ditangani menggunakan kategori `No Voucher` dan `No Referral` tanpa menghapus observasi.

```python
# DATA CLEANING - HANDLING MISSING VALUES

# Membuat salinan dataset agar data asli tetap aman
users_clean = users.copy()
transactions_clean = transactions.copy()
sellers_clean = sellers.copy()
deliveries_clean = deliveries.copy()
cs_clean = cs.copy()

# 1. Mengisi voucher yang kosong
transactions_clean["voucher_type"] = (
    transactions_clean["voucher_type"]
    .fillna("No Voucher")
)

# 2. Mengisi referral code yang kosong
users_clean["referral_code_used"] = (
    users_clean["referral_code_used"]
    .fillna("No Referral")
)

# Verifikasi hasil
print("Missing voucher_type:",
      transactions_clean["voucher_type"].isna().sum())

print("Missing referral_code_used:",
      users_clean["referral_code_used"].isna().sum())
```

```python

# DATA CLEANING - REMAINING MISSING VALUES

# Mengisi kategori yang tidak diketahui
users_clean["gender"] = (
    users_clean["gender"].fillna("Unknown")
)

deliveries_clean["courier_name"] = (
    deliveries_clean["courier_name"].fillna("Unknown")
)

# Rating tetap NaN karena tidak ada bukti nilainya
# courier_rating dan csat_score tidak diubah

# Verifikasi seluruh missing values setelah cleaning
clean_data = {
    "users": users_clean,
    "sellers": sellers_clean,
    "transactions": transactions_clean,
    "deliveries": deliveries_clean,
    "cs": cs_clean
}

for name, df in clean_data.items():
    print(f"\n{name.upper()}")
    print(df.isna().sum()[df.isna().sum() > 0])
```

### 4.4 Pemeriksaan Duplikasi Primary Key

Pemeriksaan duplikasi dilakukan untuk memastikan bahwa setiap identitas utama dalam dataset memiliki struktur yang sesuai dengan hubungan antartabel.

Meskipun pada pemeriksaan awal tidak ditemukan baris duplikat identik, kondisi tersebut belum menjamin keunikan masing-masing ID.

### 4.5 Hasil Pemeriksaan Duplikasi ID

Hasil pemeriksaan menunjukkan bahwa `user_id` pada dataset pengguna, `seller_id` pada dataset penjual, dan `ticket_id` pada dataset layanan pelanggan memiliki nilai unik.

Namun, ditemukan **1.087 kemunculan duplikat `order_id` pada dataset transaksi** dan **12 kemunculan duplikat `order_id` pada dataset pengiriman**.

Temuan ini memerlukan investigasi lebih lanjut karena duplikasi identitas transaksi berpotensi menyebabkan perhitungan nilai transaksi dan jumlah pesanan menjadi berlebih (*overcounting*).

Oleh karena itu, dilakukan pemeriksaan perbedaan informasi antarbaris dengan `order_id` yang sama sebelum menentukan metode penanganan duplikasi.

```python
# DATA CLEANING - INVESTIGASI DUPLIKASI ORDER ID

# Mengambil seluruh baris dengan order_id berulang
duplicate_transactions = transactions_clean[
    transactions_clean.duplicated(
        subset=["order_id"], keep=False
    )
].sort_values("order_id")

duplicate_deliveries = deliveries_clean[
    deliveries_clean.duplicated(
        subset=["order_id"], keep=False
    )
].sort_values("order_id")

print("=== DUPLIKASI TRANSAKSI ===")
print("Jumlah order_id bermasalah:",
      duplicate_transactions["order_id"].nunique())
print("Jumlah baris yang terlibat:",
      len(duplicate_transactions))

display(duplicate_transactions.head(10))

print("\n=== DUPLIKASI PENGIRIMAN ===")
print("Jumlah order_id bermasalah:",
      duplicate_deliveries["order_id"].nunique())
print("Jumlah baris yang terlibat:",
      len(duplicate_deliveries))

display(duplicate_deliveries.head(10))
```

```python
# ANALISIS KONSISTENSI ORDER ID BERULANG

for name, df in {
    "transactions": transactions_clean,
    "deliveries": deliveries_clean
}.items():

    duplicate_rows = df[
        df.duplicated("order_id", keep=False)
    ]

    # Hitung jumlah nilai berbeda per order_id
    comparison = (
        duplicate_rows.groupby("order_id")
        .nunique(dropna=False)
    )

    # Kolom selain order_id yang memiliki perbedaan
    varying = comparison.loc[
        :, comparison.gt(1).any()
    ]

    print(f"\n=== {name.upper()} ===")
    print("Jumlah order_id berulang:",
          duplicate_rows["order_id"].nunique())

    print("\nJumlah order_id dengan perbedaan per kolom:")
    display(varying.gt(1).sum().sort_values(ascending=False))
```

### 4.6 Investigasi Konsistensi Duplikasi

Hasil pemeriksaan lanjutan menunjukkan bahwa sebagian `order_id` berulang memiliki perbedaan informasi antarbaris, termasuk pada variabel pengiriman.

Kondisi tersebut menunjukkan adanya potensi inkonsistensi pencatatan dan risiko penggandaan observasi pada proses penggabungan dataset.

Oleh karena itu, diperlukan pemeriksaan perbedaan nilai antarbaris sebelum menetapkan metode deduplikasi. Data asli tetap dipertahankan hingga aturan penanganan dapat ditentukan secara konsisten.

### 4.7 Pemeriksaan Konsistensi Data Duplikat

Pemeriksaan lanjutan dilakukan untuk mengidentifikasi perbedaan informasi pada transaksi dan pengiriman yang memiliki `order_id` berulang.

Analisis mencakup seluruh data duplikat, bukan hanya 10 baris yang ditampilkan sebelumnya.

Tujuannya adalah menentukan apakah duplikasi merupakan kesalahan pencatatan atau observasi yang perlu dipertahankan, sehingga proses pembersihan data tidak menyebabkan hilangnya informasi transaksi yang valid.

```python
# PEMERIKSAAN PERBEDAAN DATA PADA ORDER ID BERULANG

for name, df in {
    "transactions": transactions_clean,
    "deliveries": deliveries_clean
}.items():

    duplicates = df[
        df.duplicated("order_id", keep=False)
    ]

    comparison = (
        duplicates.groupby("order_id")
        .nunique(dropna=False)
    )

    # Menghitung order_id dengan nilai berbeda per kolom
    differences = comparison.gt(1).sum()
    differences = differences[
        differences > 0
    ].sort_values(ascending=False)

    print(f"\n=== {name.upper()} ===")
    print("Jumlah order_id berulang:",
          duplicates["order_id"].nunique())

    print("Jumlah baris terlibat:", len(duplicates))

    print("\nKolom yang memiliki perbedaan:")
    display(differences.to_frame(
        name="Jumlah Order ID Berbeda"
    ))
```

### 4.8 Validasi Identitas Transaksi

Hasil investigasi menunjukkan adanya perbedaan informasi pada sejumlah `order_id` berulang, baik dalam dataset transaksi maupun pengiriman.

Untuk menentukan metode penanganan yang tepat, dilakukan pemeriksaan terhadap `user_id`, `order_date`, dan `order_status` pada transaksi yang memiliki identitas pesanan sama.

Pemeriksaan ini bertujuan membedakan potensi benturan ID dari duplikasi pencatatan sebelum melakukan pembersihan data.

```python
# VALIDASI IDENTITAS ORDER ID BERULANG

dup_tx = transactions_clean[
    transactions_clean.duplicated("order_id", keep=False)
]

check_identity = (
    dup_tx.groupby("order_id")
    .agg(
        jumlah_baris=("order_id", "size"),
        jumlah_user=("user_id", "nunique"),
        jumlah_tanggal=("order_date", "nunique"),
        jumlah_status=("order_status", "nunique")
    )
)

print("Total order_id berulang:", len(check_identity))

print("\nOrder ID dengan user berbeda:",
      (check_identity["jumlah_user"] > 1).sum())

print("Order ID dengan tanggal berbeda:",
      (check_identity["jumlah_tanggal"] > 1).sum())

print("Order ID dengan status berbeda:",
      (check_identity["jumlah_status"] > 1).sum())

display(check_identity.head(10))
```

### 4.9 Identifikasi Penyebab Duplikasi Transaksi

Hasil pemeriksaan menunjukkan 1.054 `order_id` berulang, dengan 16 ID terkait pengguna berbeda dan dua ID memiliki status transaksi berbeda.

Untuk menentukan tindakan cleaning yang tepat, dilakukan pemeriksaan karakteristik transaksi pada `order_id` berulang, khususnya perbedaan kategori produk, penjual, jumlah item, dan nilai transaksi.

Pemeriksaan ini bertujuan menilai apakah pengulangan ID berkaitan dengan pencatatan beberapa item dalam satu pesanan atau menunjukkan inkonsistensi data.

```python
# INVESTIGASI STRUKTUR TRANSAKSI BERULANG

dup_tx = transactions_clean[
    transactions_clean.duplicated("order_id", keep=False)
]

item_check = (
    dup_tx.groupby("order_id")
    .agg(
        jumlah_baris=("order_id", "size"),
        jumlah_user=("user_id", "nunique"),
        jumlah_seller=("seller_id", "nunique"),
        jumlah_kategori=("product_category", "nunique"),
        jumlah_qty=("item_quantity", "nunique"),
        jumlah_nominal=("total_amount", "nunique")
    )
)

print("Order dengan seller berbeda:",
      (item_check["jumlah_seller"] > 1).sum())

print("Order dengan kategori berbeda:",
      (item_check["jumlah_kategori"] > 1).sum())

print("Order dengan quantity berbeda:",
      (item_check["jumlah_qty"] > 1).sum())

print("Order dengan total_amount berbeda:",
      (item_check["jumlah_nominal"] > 1).sum())

display(item_check.head(10))
```

### 4.10 Klasifikasi Duplikasi Transaksi

Hasil pemeriksaan menunjukkan adanya `order_id` berulang dengan perbedaan informasi pada sejumlah variabel, termasuk identitas pengguna, penjual, kategori produk, dan nilai transaksi.

Untuk menentukan tindakan pembersihan yang tepat, data duplikat diklasifikasikan menjadi dua kelompok: pencatatan berulang dengan informasi bisnis yang sama dan pencatatan dengan informasi bisnis yang berbeda.

Klasifikasi ini dilakukan agar proses deduplikasi tidak menyebabkan penghapusan transaksi valid maupun penggandaan nilai transaksi.

```python

# KLASIFIKASI DUPLIKASI TRANSAKSI

# transaction_key merupakan ID teknis baris
kolom_bisnis = [
    col for col in transactions_clean.columns
    if col not in ["transaction_key", "order_id"]
]

# Hitung variasi nilai untuk setiap order_id berulang
variasi = (
    transactions_clean[
        transactions_clean.duplicated("order_id", keep=False)
    ]
    .groupby("order_id")[kolom_bisnis]
    .nunique(dropna=False)
)

# Kelompokkan berdasarkan konsistensi data
duplikat_sama = variasi.index[
    variasi.eq(1).all(axis=1)
]

duplikat_berbeda = variasi.index[
    variasi.gt(1).any(axis=1)
]

print("=== HASIL KLASIFIKASI ===")
print("Order ID dengan informasi bisnis sama:",
      len(duplikat_sama))

print("Order ID dengan informasi bisnis berbeda:",
      len(duplikat_berbeda))

print("Total order ID berulang:",
      len(variasi))

# Ringkasan perbedaan kolom untuk kelompok konflik
display(
    variasi.loc[duplikat_berbeda]
    .gt(1).sum()
    .sort_values(ascending=False)
    .to_frame("Jumlah Order ID Berbeda")
)
```

### 4.11 Penanganan Duplikasi Transaksi

Hasil pemeriksaan terhadap 1.054 `order_id` berulang menunjukkan bahwa 1.038 ID memiliki informasi bisnis yang konsisten, sedangkan 16 ID memiliki perbedaan informasi.

Untuk mencegah penggandaan nilai transaksi, observasi berulang dengan informasi bisnis identik akan dideduplikasi dengan mempertahankan satu baris per `order_id`.

Sementara itu, 16 `order_id` dengan informasi bertentangan akan dipisahkan untuk ditinjau lebih lanjut dan tidak langsung digabungkan dalam analisis utama berbasis pesanan. Pendekatan ini bertujuan menjaga validitas perhitungan jumlah transaksi, nilai promosi, dan retensi pelanggan.

```python

# PENANGANAN DUPLIKASI TRANSAKSI

# Pisahkan seluruh baris dengan order_id konflik
transactions_conflict = transactions_clean[
    transactions_clean["order_id"].isin(
        duplikat_berbeda
    )
].copy()

# Keluarkan order_id konflik dari data utama
transactions_valid = transactions_clean[
    ~transactions_clean["order_id"].isin(
        duplikat_berbeda
    )
].copy()

# Hilangkan pencatatan berulang pada order_id konsisten
transactions_valid = (
    transactions_valid
    .drop_duplicates(subset=["order_id"], keep="first")
    .copy()
)

# Validasi hasil
print("Total baris awal:", len(transactions_clean))
print("Baris transaksi valid:", len(transactions_valid))
print("Baris konflik dipisahkan:", len(transactions_conflict))
print("ID unik setelah cleaning:",
      transactions_valid["order_id"].nunique())
print("Duplikat ID tersisa:",
      transactions_valid["order_id"].duplicated().sum())
```

## 5. Feature Engineering

### 5.1 Perhitungan Repeat Order Rate 90 Hari

Repeat Order Rate 90 hari digunakan untuk mengukur persentase pelanggan yang melakukan pembelian kedua dalam waktu maksimal 90 hari sejak pembelian pertama yang teramati.

Perhitungan menggunakan transaksi berstatus `completed` selama 2023–2024. Untuk menghindari bias, hanya pelanggan dengan periode observasi minimal 90 hari setelah pembelian pertama yang dimasukkan dalam perhitungan utama.

Pengguna dengan pembelian pertama setelah 2 Oktober 2024 tidak dimasukkan dalam denominator karena belum memiliki waktu observasi 90 hari penuh hingga akhir dataset.

*Catatan: Pembelian pertama dalam analisis merupakan pembelian pertama yang tercatat dalam periode penelitian, bukan selalu pembelian pertama sepanjang riwayat pelanggan.*

```python

# FEATURE ENGINEERING - REPEAT ORDER RATE 90 HARI

# 1. Urutkan transaksi berdasarkan pelanggan dan tanggal
orders = transactions_completed.sort_values(
    ["user_id", "order_date", "order_id"]
).copy()

# 2. Berikan nomor urutan transaksi tiap pelanggan
orders["order_number"] = (
    orders.groupby("user_id").cumcount() + 1
)

# 3. Ambil tanggal transaksi pertama
first_orders = orders[
    orders["order_number"] == 1
][["user_id", "order_date"]].rename(
    columns={"order_date": "first_order_date"}
)

# 4. Ambil tanggal transaksi kedua
second_orders = orders[
    orders["order_number"] == 2
][["user_id", "order_date"]].rename(
    columns={"order_date": "second_order_date"}
)

# 5. Gabungkan transaksi pertama dan kedua
retention = first_orders.merge(
    second_orders,
    on="user_id",
    how="left",
    validate="one_to_one"
)

# 6. Hitung jarak hari pembelian pertama dan kedua
retention["days_to_second_order"] = (
    retention["second_order_date"]
    - retention["first_order_date"]
).dt.days

# 7. Tentukan pelanggan dengan observasi minimal 90 hari
observation_end = pd.Timestamp("2024-12-31")

retention["eligible_90d"] = (
    retention["first_order_date"]
    <= observation_end - pd.Timedelta(days=90)
)

# 8. Tentukan apakah pelanggan repeat dalam 90 hari
retention["repeat_90d"] = (
    retention["days_to_second_order"].between(0, 90)
)

# 9. Filter pelanggan yang memenuhi periode observasi
retention_eligible = retention[
    retention["eligible_90d"]
].copy()

# 10. Hitung Repeat Order Rate
repeat_rate_90d = (
    retention_eligible["repeat_90d"].mean() * 100
)

print("=== HASIL ANALISIS RETENSI ===")
print("Total pengguna:", len(retention))
print("Pengguna eligible:", len(retention_eligible))
print("Pengguna repeat 90 hari:",
      retention_eligible["repeat_90d"].sum())
print(f"Repeat Order Rate 90 Hari: {repeat_rate_90d:.2f}%")

display(retention.head(10))
```

```python
# VALIDASI TAMBAHAN REPEAT ORDER RATE 90 HARI

print("=== VALIDASI RETENSI ===")

print("1. Total pengguna eligible:",
      len(retention_eligible))

print("2. Jumlah pengguna unik:",
      retention_eligible["user_id"].nunique())

print("3. Pengguna repeat:",
      retention_eligible["repeat_90d"].sum())

print("4. Pengguna tidak repeat:",
      (~retention_eligible["repeat_90d"]).sum())

print("5. Repeat Order Rate:",
      round(retention_eligible["repeat_90d"].mean() * 100, 2), "%")

print("6. Duplikasi user_id:",
      retention_eligible["user_id"].duplicated().sum())

print("7. Transaksi kedua sebelum transaksi pertama:",
      (retention_eligible["days_to_second_order"] < 0).sum())

print("8. Tanggal transaksi pertama paling akhir:",
      retention_eligible["first_order_date"].max())
```

```python
# VALIDASI TREN RETENSI PER TAHUN

retention_by_year = (
    retention_eligible
    .assign(
        year=retention_eligible["first_order_date"].dt.year
    )
    .groupby("year")
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum"),
        repeat_rate=("repeat_90d", "mean")
    )
    .reset_index()
)

retention_by_year["repeat_rate"] *= 100

display(retention_by_year.round(2))
```

### 5.3 Cohort Retention Analysis

Analisis cohort dilakukan untuk membandingkan tingkat pembelian ulang pelanggan berdasarkan periode transaksi pertama yang teramati.

Setiap pengguna dikelompokkan berdasarkan bulan pembelian pertama, kemudian dihitung persentase pengguna yang melakukan pembelian kedua dalam 90 hari.

```python
# COHORT RETENTION ANALYSIS

# Gunakan hanya pengguna dengan observasi 90 hari penuh
cohort_data = retention_eligible.copy()

# Membuat cohort berdasarkan bulan pembelian pertama
cohort_data["cohort_month"] = (
    cohort_data["first_order_date"]
    .dt.to_period("M")
)

# Menghitung repeat rate setiap cohort
cohort_summary = (
    cohort_data.groupby("cohort_month")
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum")
    )
    .reset_index()
)

cohort_summary["repeat_rate_90d"] = (
    cohort_summary["repeat_users"]
    / cohort_summary["total_users"] * 100
).round(2)

cohort_summary["cohort_month"] = (
    cohort_summary["cohort_month"].astype(str)
)

display(cohort_summary)
```

## 6. Analisis Efisiensi Promosi

### 6.1 Hubungan Intensitas Promo dan Repeat Order

Analisis ini bertujuan mengidentifikasi hubungan antara besarnya insentif pada transaksi pertama dan kemungkinan pelanggan melakukan pembelian kedua dalam 90 hari.

Intensitas promo dihitung sebagai proporsi `promo_amount` terhadap `total_amount`, dengan asumsi kedua nominal memiliki definisi yang sesuai untuk dibandingkan.

Pengguna kemudian dikelompokkan berdasarkan intensitas promo pada transaksi pertama, lalu dibandingkan berdasarkan Repeat Order Rate 90 hari.

Analisis ini bersifat observasional sehingga hubungan yang ditemukan tidak secara langsung menunjukkan pengaruh sebab-akibat.

```python
# ANALISIS PROMO VS REPEAT ORDER

# Ambil promo dari transaksi pertama setiap pelanggan
first_order_promo = orders[
    orders["order_number"] == 1
][["user_id", "total_amount", "promo_amount"]].copy()

# Gabungkan dengan pengguna eligible
promo_retention = retention_eligible.merge(
    first_order_promo,
    on="user_id",
    how="left",
    validate="one_to_one"
)

# Hitung promo intensity
promo_retention["promo_intensity"] = np.where(
    promo_retention["total_amount"] > 0,
    promo_retention["promo_amount"]
    / promo_retention["total_amount"],
    np.nan
)

# Periksa nilai intensity sebelum segmentasi
display(
    promo_retention["promo_intensity"]
    .describe(percentiles=[.25, .5, .75, .9, .95, .99])
)

print("Nilai promo intensity negatif:",
      (promo_retention["promo_intensity"] < 0).sum())

print("Nilai promo intensity > 100%:",
      (promo_retention["promo_intensity"] > 1).sum())

print("Total amount <= 0:",
      (promo_retention["total_amount"] <= 0).sum())
```

### 6.2 Validasi Intensitas Promo

Hasil pemeriksaan menunjukkan bahwa dari 47.350 pengguna eligible, sebanyak 47.285 memiliki rasio promo yang dapat dihitung.

Ditemukan 2.116 transaksi dengan rasio promo melebihi 100% serta 65 transaksi dengan `total_amount` kurang dari atau sama dengan nol.

Rata-rata intensitas promo sebesar 23,45% jauh lebih tinggi dibandingkan median 0,77%, menunjukkan distribusi yang sangat menceng ke kanan akibat nilai ekstrem.

Kondisi tersebut memerlukan pemeriksaan definisi `total_amount` dan `promo_amount` sebelum rasio digunakan untuk segmentasi efisiensi promosi.

```python

# INVESTIGASI ANOMALI INTENSITAS PROMO

# Ambil transaksi pertama yang memiliki rasio promo > 100%
promo_anomaly = promo_retention[
    promo_retention["promo_intensity"] > 1
].copy()

print("Jumlah anomali:", len(promo_anomaly))

# Distribusi nominal transaksi dan promo
display(
    promo_anomaly[
        ["total_amount", "promo_amount", "promo_intensity"]
    ].describe().round(2)
)

# Menampilkan contoh transaksi dengan intensitas terbesar
display(
    promo_anomaly[
        ["user_id", "total_amount",
         "promo_amount", "promo_intensity"]
    ].sort_values(
        "promo_intensity", ascending=False
    ).head(10)
)

# Pemeriksaan nilai total_amount <= 0
print("\nTransaksi dengan total_amount <= 0:")

display(
    promo_retention.loc[
        promo_retention["total_amount"] <= 0,
        ["user_id", "total_amount", "promo_amount"]
    ].head(10)
)
```

### 6.3 Penyesuaian Metode Pengukuran Promo

Hasil investigasi menemukan 2.116 transaksi dengan rasio promo terhadap nilai transaksi melebihi 100%, serta 65 transaksi dengan nilai transaksi nol atau negatif.

Kondisi ini menunjukkan bahwa rasio `promo_amount / total_amount` belum dapat digunakan secara langsung sebagai ukuran efisiensi promosi tanpa validasi lebih lanjut terhadap definisi nominal transaksi.

Oleh karena itu, analisis segmentasi awal menggunakan nominal promo absolut pada transaksi pertama. Pendekatan ini memungkinkan perbandingan retensi berdasarkan besarnya insentif tanpa dipengaruhi denominator yang tidak stabil.

Meskipun demikian, nominal promo absolut belum dapat diinterpretasikan sebagai ukuran efisiensi promosi secara menyeluruh karena belum memperhitungkan nilai transaksi, margin, dan faktor karakteristik pelanggan.

```python

# SEGMENTASI PELANGGAN BERDASARKAN NOMINAL PROMO

promo_retention["promo_segment"] = pd.cut(
    promo_retention["promo_amount"],
    bins=[-0.01, 0, 25000, 100000, float("inf")],
    labels=[
        "Tanpa Promo",
        "Promo Rendah (1-25rb)",
        "Promo Sedang (25-100rb)",
        "Promo Tinggi (>100rb)"
    ],
    include_lowest=True
)

# Menghitung Repeat Order Rate setiap segmen
promo_segment_summary = (
    promo_retention.groupby(
        "promo_segment", observed=True
    )
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum"),
        avg_promo=("promo_amount", "mean"),
        total_promo=("promo_amount", "sum")
    )
    .reset_index()
)

promo_segment_summary["repeat_rate_90d"] = (
    promo_segment_summary["repeat_users"]
    / promo_segment_summary["total_users"] * 100
).round(2)

display(promo_segment_summary)
```

### 6.4 Hasil Analisis Segmentasi Promo dan Retensi

Hasil segmentasi menunjukkan bahwa Repeat Order Rate 90 hari relatif serupa pada seluruh kelompok promo, yaitu berkisar antara 76,03% hingga 77,73%.

Kelompok Promo Tinggi memiliki rata-rata insentif sekitar Rp750.272, tetapi tingkat retensinya hanya 0,53 poin persentase lebih tinggi dibandingkan kelompok Tanpa Promo.

Temuan ini mengindikasikan adanya potensi ketidakefisienan alokasi promosi, karena pemberian insentif yang jauh lebih besar tidak diikuti perbedaan tingkat retensi yang sebanding.

Namun, hubungan tersebut bersifat observasional dan masih perlu dianalisis berdasarkan karakteristik pelanggan, nilai transaksi, serta kanal akuisisi sebelum menghasilkan rekomendasi final.

### 6.5 Evaluasi Alokasi Promo per Pengguna Repeat

Untuk membandingkan penggunaan insentif antarsegmen, dihitung rasio total promo pada transaksi pertama terhadap jumlah pengguna yang melakukan pembelian kedua dalam 90 hari.

Indikator ini digunakan sebagai ukuran deskriptif intensitas pengeluaran promo relatif terhadap jumlah pengguna repeat, bukan sebagai estimasi biaya kausal untuk menghasilkan retensi.

Hasilnya digunakan untuk mengidentifikasi segmen yang perlu diprioritaskan dalam evaluasi alokasi budget promosi.

```python

# ANALISIS PROMO PER REPEAT USER

promo_segment_summary["promo_per_repeat_user"] = (
    promo_segment_summary["total_promo"]
    / promo_segment_summary["repeat_users"]
)

display(
    promo_segment_summary[
        [
            "promo_segment",
            "total_users",
            "repeat_users",
            "total_promo",
            "repeat_rate_90d",
            "promo_per_repeat_user"
        ]
    ].round(2)
)

# Persentase alokasi promo antarsegmen
promo_segment_summary["promo_share_pct"] = (
    promo_segment_summary["total_promo"]
    / promo_segment_summary["total_promo"].sum()
    * 100
).round(2)

display(
    promo_segment_summary[
        ["promo_segment", "promo_share_pct"]
    ]
)
```

## 7. Analisis Kanal Akuisisi Pelanggan

### 7.1 Hubungan Kanal Akuisisi, Promosi, dan Retensi

Analisis ini bertujuan membandingkan perilaku pembelian ulang pelanggan berdasarkan kanal akuisisi, yaitu `ads`, `organic`, `affiliate`, dan `referral`.

Evaluasi dilakukan menggunakan tingkat repeat order 90 hari, nominal promo transaksi pertama, serta distribusi pelanggan pada masing-masing kanal.

Tujuannya adalah mengidentifikasi kanal dengan tingkat retensi yang relatif rendah meskipun menerima alokasi promosi yang tinggi.

Analisis bersifat deskriptif karena dataset tidak menyediakan biaya akuisisi pelanggan (CAC) aktual untuk setiap kanal.

```python

# ANALISIS AKUISISI, PROMO, DAN RETENSI

# 1. Menggabungkan data pengguna
acquisition_analysis = promo_retention.merge(
    users_clean[
        ["user_id", "acquisition_channel"]
    ],
    on="user_id",
    how="left",
    validate="one_to_one"
)

# 2. Memeriksa hasil penggabungan
print("Total pengguna sebelum merge:", len(promo_retention))
print("Total pengguna setelah merge:", len(acquisition_analysis))
print("Kanal akuisisi kosong:",
      acquisition_analysis["acquisition_channel"].isna().sum())

# 3. Ringkasan per kanal akuisisi
acquisition_summary = (
    acquisition_analysis.groupby(
        "acquisition_channel", dropna=False
    )
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum"),
        total_promo=("promo_amount", "sum"),
        avg_promo=("promo_amount", "mean")
    )
    .reset_index()
)

# 4. Hitung tingkat retensi
acquisition_summary["repeat_rate_90d"] = (
    acquisition_summary["repeat_users"]
    / acquisition_summary["total_users"] * 100
).round(2)

# 5. Hitung nominal promo per pengguna repeat
acquisition_summary["promo_per_repeat_user"] = (
    acquisition_summary["total_promo"]
    / acquisition_summary["repeat_users"].replace(0, np.nan)
).round(2)

# 6. Persentase alokasi promo
acquisition_summary["promo_share_pct"] = (
    acquisition_summary["total_promo"]
    / acquisition_summary["total_promo"].sum()
    * 100
).round(2)

# 7. Urutkan berdasarkan total promo
acquisition_summary = acquisition_summary.sort_values(
    "total_promo", ascending=False
)

display(acquisition_summary)
```

## 8. Analisis Customer Experience dan Retensi

### 8.1 Pengaruh Pengalaman Pengiriman dan Keluhan Pelanggan

Analisis ini bertujuan mengidentifikasi hubungan antara pengalaman pelanggan pada transaksi pertama dan pembelian ulang dalam 90 hari.

Variabel yang digunakan mencakup keterlambatan pengiriman, status pengiriman, serta keberadaan keluhan pelanggan.

Untuk menjaga urutan waktu analisis, informasi layanan hanya digunakan apabila terjadi sebelum transaksi kedua atau dalam periode pengamatan yang sesuai. Hubungan yang ditemukan bersifat observasional dan tidak secara langsung membuktikan sebab-akibat.

```python

# CUSTOMER EXPERIENCE - PERSIAPAN DATA

# 1. Ambil transaksi pertama pengguna eligible
first_order_experience = orders[
    orders["order_number"] == 1
][["user_id", "order_id", "order_date"]].merge(
    retention_eligible[
        ["user_id", "second_order_date", "repeat_90d"]
    ],
    on="user_id",
    how="inner",
    validate="one_to_one"
)

# 2. Ambil informasi pengiriman
delivery_info = deliveries_valid[
    ["order_id", "dispatch_date", "delivery_date",
     "delivery_status", "sla_promised_days"]
].copy()

delivery_info["delivery_days"] = (
    delivery_info["delivery_date"]
    - delivery_info["dispatch_date"]
).dt.days

delivery_info["is_late"] = (
    delivery_info["delivery_days"]
    > delivery_info["sla_promised_days"]
)

# 3. Gabungkan dengan transaksi pertama
experience = first_order_experience.merge(
    delivery_info,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# 4. Tandai pengiriman dengan informasi waktu valid
experience["valid_delivery"] = (
    experience["dispatch_date"].notna()
    & experience["delivery_date"].notna()
    & (experience["delivery_days"] >= 0)
    & (experience["sla_promised_days"] >= 0)
    & (experience["delivery_date"] >= experience["order_date"])
)

# 5. Hanya gunakan pengiriman yang selesai
# sebelum pembelian kedua, bila ada
experience["delivery_before_repeat"] = (
    experience["second_order_date"].isna()
    | (
        experience["delivery_date"]
        <= experience["second_order_date"]
    )
)

experience["eligible_delivery_analysis"] = (
    experience["valid_delivery"]
    & experience["delivery_before_repeat"]
)

print("Total pengguna eligible:", len(experience))
print("Data pengiriman tersedia:",
      experience["delivery_status"].notna().sum())
print("Pengiriman memenuhi kriteria analisis:",
      experience["eligible_delivery_analysis"].sum())

print("\nDistribusi status pengiriman:")
display(experience["delivery_status"].value_counts(dropna=False))
```

### 8.2 Validasi Data Pengiriman

Dari 47.350 pengguna yang memenuhi kriteria analisis retensi, seluruhnya memiliki catatan pengiriman yang terhubung melalui transaksi pertama.

Sebanyak 42.205 catatan memenuhi kriteria kelengkapan dan konsistensi waktu yang ditetapkan.

Status pengiriman terdiri dari 43.017 `delivered`, 2.133 `failed`, 1.274 `returned`, dan 926 `in_transit`.

Untuk analisis keterlambatan, hanya pengiriman berstatus `delivered` dengan data tanggal yang valid yang akan digunakan. Pendekatan ini bertujuan menghindari kesalahan klasifikasi pengiriman yang belum selesai atau gagal.

```python

# ANALISIS KETERLAMBATAN PENGIRIMAN VS RETENSI

# Hanya pengiriman delivered yang valid
delivery_analysis = experience[
    (experience["delivery_status"] == "delivered")
    & (experience["eligible_delivery_analysis"])
].copy()

# Klasifikasi ketepatan waktu
delivery_analysis["delivery_performance"] = np.where(
    delivery_analysis["is_late"],
    "Terlambat",
    "Tepat Waktu"
)

# Hitung repeat rate berdasarkan ketepatan pengiriman
delivery_retention_summary = (
    delivery_analysis.groupby("delivery_performance")
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum"),
        avg_delivery_days=("delivery_days", "mean")
    )
    .reset_index()
)

delivery_retention_summary["repeat_rate_90d"] = (
    delivery_retention_summary["repeat_users"]
    / delivery_retention_summary["total_users"] * 100
).round(2)

display(delivery_retention_summary)

print("Total pengguna dianalisis:", len(delivery_analysis))
```

### 8.3 Hasil Analisis Keterlambatan Pengiriman

Dari 38.242 pengguna yang memenuhi kriteria analisis pengiriman, sebanyak 25.659 pengguna menerima pengiriman tepat waktu, sedangkan 12.583 pengguna mengalami keterlambatan.

Repeat Order Rate 90 hari pada kelompok pengiriman tepat waktu tercatat sebesar 81,87%, dibandingkan 67,05% pada kelompok pengiriman terlambat, dengan selisih 14,82 poin persentase.

Hasil ini menunjukkan adanya hubungan antara ketepatan pengiriman dan retensi pelanggan dalam sampel yang dianalisis.

Namun, perbandingan tersebut masih bersifat eksploratif karena pemilihan sampel berdasarkan waktu pengiriman dan transaksi kedua berpotensi menimbulkan bias. Oleh karena itu, diperlukan pemeriksaan sensitivitas sebelum hasil digunakan sebagai dasar rekomendasi final.

```python

# SENSITIVITY CHECK - DELIVERY VS RETENTION

delivery_sensitivity = experience[
    (experience["delivery_status"] == "delivered")
    & (experience["valid_delivery"])
    & (experience["delivery_date"] <= pd.Timestamp("2024-12-31"))
].copy()

delivery_sensitivity["delivery_performance"] = np.where(
    delivery_sensitivity["is_late"],
    "Terlambat",
    "Tepat Waktu"
)

delivery_sensitivity_summary = (
    delivery_sensitivity.groupby("delivery_performance")
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum")
    )
    .reset_index()
)

delivery_sensitivity_summary["repeat_rate_90d"] = (
    delivery_sensitivity_summary["repeat_users"]
    / delivery_sensitivity_summary["total_users"] * 100
).round(2)

display(delivery_sensitivity_summary)

print("Total pengguna:", len(delivery_sensitivity))
```

### 8.4 Analisis Keluhan Pelanggan dan Retensi

Analisis ini bertujuan mengidentifikasi hubungan antara keluhan pelanggan pada transaksi pertama dan kemungkinan pembelian ulang.

Keluhan diamati selama 14 hari pertama setelah pembelian, sedangkan pembelian ulang dievaluasi pada hari ke-15 hingga ke-90. Pemisahan periode ini digunakan untuk menjaga urutan waktu antara keluhan dan pembelian ulang.

Analisis dilakukan berdasarkan keberadaan tiket Customer Service, bukan sebagai bukti hubungan sebab-akibat.

```python

# CUSTOMER SERVICE VS CUSTOMER RETENTION

# 1. Data transaksi pertama pelanggan eligible
cs_base = first_order_experience[
    ["user_id", "order_id", "order_date"]
].copy()

# 2. Hubungkan tiket CS dengan transaksi pertama
cs_tickets_first = cs_clean.merge(
    cs_base,
    on="order_id",
    how="inner",
    validate="many_to_one"
)

# 3. Hitung jarak tanggal tiket dari transaksi pertama
cs_tickets_first["days_to_ticket"] = (
    cs_tickets_first["ticket_date"]
    - cs_tickets_first["order_date"]
).dt.days

# 4. Ambil tiket dalam 14 hari pertama
cs_14d = cs_tickets_first[
    cs_tickets_first["days_to_ticket"].between(0, 14)
].copy()

ticket_counts = (
    cs_14d.groupby("user_id")
    .size()
    .rename("jumlah_tiket_14d")
    .reset_index()
)

# 5. Bentuk status keluhan per pengguna
cs_analysis = cs_base.merge(
    ticket_counts,
    on="user_id",
    how="left",
    validate="one_to_one"
)

cs_analysis["jumlah_tiket_14d"] = (
    cs_analysis["jumlah_tiket_14d"].fillna(0).astype(int)
)

cs_analysis["complaint_group"] = np.where(
    cs_analysis["jumlah_tiket_14d"] > 0,
    "Ada Tiket CS",
    "Tidak Ada Tiket CS"
)

# 6. Hitung repeat order pada hari ke-15 hingga ke-90
orders_with_first = orders.merge(
    first_orders,
    on="user_id",
    how="left",
    validate="many_to_one"
)

orders_with_first["days_since_first"] = (
    orders_with_first["order_date"]
    - orders_with_first["first_order_date"]
).dt.days

repeat_ids = orders_with_first.loc[
    orders_with_first["days_since_first"].between(15, 90),
    "user_id"
].unique()

cs_analysis["repeat_15_90d"] = (
    cs_analysis["user_id"].isin(repeat_ids)
)

# 7. Ringkasan hasil
cs_summary = (
    cs_analysis.groupby("complaint_group")
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_15_90d", "sum"),
        total_tickets=("jumlah_tiket_14d", "sum")
    )
    .reset_index()
)

cs_summary["repeat_rate_15_90d"] = (
    cs_summary["repeat_users"]
    / cs_summary["total_users"] * 100
).round(2)

display(cs_summary)

print("Total pengguna dianalisis:", len(cs_analysis))
print("Total tiket dalam 14 hari:", len(cs_14d))
```

### 8.5 Hasil Analisis Customer Service dan Retensi

Dari 47.350 pengguna yang dianalisis, terdapat 4.332 pengguna dengan tiket Customer Service dalam 14 hari pertama setelah transaksi.

### 11.5 Customer Lifetime Value & Promo Efficiency Analysis

Analisis ini bertujuan mengevaluasi apakah alokasi promosi pada transaksi pertama berkaitan dengan nilai pelanggan yang berkelanjutan, bukan hanya mendorong pembelian awal.

Dengan menggunakan data transaksi completed, nilai pelanggan dihitung berdasarkan total nilai pembelian dan frekuensi transaksi selama 90 hari sejak transaksi pertama yang teramati.

Analisis dilakukan pada pelanggan dengan periode observasi yang memadai dan dibandingkan berdasarkan segmen nominal promo transaksi pertama.

Karena dataset belum menyediakan margin kontribusi aktual dan riwayat pelanggan sepanjang masa hidupnya, hasil disajikan sebagai Observed Customer Value 90 Hari, bukan Customer Lifetime Value (CLV) aktual. Estimasi CLV berbasis asumsi margin dapat dikembangkan sebagai analisis skenario terpisah.

```python

# 11.5 CUSTOMER VALUE ANALYSIS - 90 HARI

import pandas as pd
import numpy as np

# 1. Ambil pelanggan dengan observasi 90 hari
clv_users = retention_eligible[
    ["user_id", "first_order_date", "repeat_90d"]
].copy()

clv_users["first_order_date"] = pd.to_datetime(
    clv_users["first_order_date"]
)

# 2. Ambil transaksi completed
clv_orders = transactions_completed[
    ["user_id", "order_id", "order_date",
     "total_amount", "promo_amount"]
].copy()

clv_orders["order_date"] = pd.to_datetime(
    clv_orders["order_date"]
)

# 3. Gabungkan transaksi dengan tanggal awal pelanggan
clv_orders = clv_orders.merge(
    clv_users[["user_id", "first_order_date"]],
    on="user_id",
    how="inner",
    validate="many_to_one"
)

# 4. Batasi transaksi dalam 90 hari
clv_orders["days_since_first"] = (
    clv_orders["order_date"]
    - clv_orders["first_order_date"]
).dt.days

clv_90d_orders = clv_orders[
    clv_orders["days_since_first"].between(0, 90)
].copy()

# 5. Tandai transaksi pertama yang teramati
first_orders = (
    clv_90d_orders
    .sort_values(["user_id", "order_date", "order_id"])
    .drop_duplicates("user_id", keep="first")
    [["user_id", "order_id", "promo_amount"]]
    .rename(columns={
        "order_id": "first_order_id",
        "promo_amount": "first_order_promo"
    })
)

# 6. Agregasi nilai pelanggan selama 90 hari
customer_value = (
    clv_90d_orders
    .groupby("user_id")
    .agg(
        orders_90d=("order_id", "nunique"),
        customer_value_90d=("total_amount", "sum")
    )
    .reset_index()
)

customer_value = customer_value.merge(
    first_orders,
    on="user_id",
    how="left",
    validate="one_to_one"
)

# 7. Gabungkan dengan data retention
customer_value = clv_users.merge(
    customer_value,
    on="user_id",
    how="left",
    validate="one_to_one"
)

# 8. Segmentasi berdasarkan promo pertama
def classify_promo(x):
    if pd.isna(x):
        return "Tidak Diketahui"
    elif x == 0:
        return "Tanpa Promo"
    elif 0 < x <= 25000:
        return "Promo Rendah (1-25rb)"
    elif 25000 < x <= 100000:
        return "Promo Sedang (25-100rb)"
    elif x > 100000:
        return "Promo Tinggi (>100rb)"
    return "Anomali"

customer_value["promo_segment"] = (
    customer_value["first_order_promo"]
    .apply(classify_promo)
)

# 9. Ringkasan per segmen
clv_segment_summary = (
    customer_value
    .groupby("promo_segment", dropna=False)
    .agg(
        total_users=("user_id", "nunique"),
        repeat_users=("repeat_90d", "sum"),
        avg_orders_90d=("orders_90d", "mean"),
        avg_value_90d=("customer_value_90d", "mean"),
        total_value_90d=("customer_value_90d", "sum"),
        avg_first_promo=("first_order_promo", "mean"),
        total_first_promo=("first_order_promo", "sum")
    )
    .reset_index()
)

clv_segment_summary["repeat_rate_90d"] = (
    clv_segment_summary["repeat_users"]
    / clv_segment_summary["total_users"] * 100
)

clv_segment_summary["value_to_promo_ratio"] = (
    clv_segment_summary["total_value_90d"]
    / clv_segment_summary["total_first_promo"]
      .replace(0, np.nan)
)

# 10. Pemeriksaan kualitas data
print("=== VALIDASI CUSTOMER VALUE ===")
print("Eligible users:", len(clv_users))
print("Customer rows:", len(customer_value))
print("Users tanpa order 90 hari:",
      customer_value["orders_90d"].isna().sum())
print("Nilai transaksi negatif:",
      (clv_90d_orders["total_amount"] < 0).sum())
print("Promo pertama negatif:",
      (customer_value["first_order_promo"] < 0).sum())
print("Duplikat user:",
      customer_value["user_id"].duplicated().sum())

print("\n=== CUSTOMER VALUE 90 HARI ===")
display(clv_segment_summary.round(2))
```

Hasil analisis menunjukkan bahwa segmen Promo Sedang memiliki rata-rata nilai transaksi pelanggan 90 hari tertinggi, yaitu Rp8,26 juta, sedangkan segmen Promo Tinggi mencatat sekitar Rp7,25 juta.

Meskipun segmen Promo Tinggi memiliki Repeat Order Rate tertinggi sebesar 77,73%, tingkat tersebut hanya sedikit lebih tinggi dibandingkan kelompok Tanpa Promo sebesar 77,20%.

Pendekatan kedua hanya mengalikan nilai transaksi dengan asumsi margin, untuk menghindari potensi pengurangan promo dua kali ketika nilai transaksi telah dicatat setelah diskon.

Hasil kedua pendekatan merupakan proxy kontribusi 90 hari, bukan CLV aktual. Biaya akuisisi lainnya, biaya operasional, biaya promosi setelah transaksi pertama, serta proyeksi nilai pelanggan setelah 90 hari belum dimasukkan.

```python

# 11.5.3 CLV PROXY - ANALISIS SENSITIVITAS MARGIN

import pandas as pd
import numpy as np

clv_analysis = customer_value.copy()

# Nilai hilang harus diperiksa, bukan dianggap nol
assert clv_analysis["customer_value_90d"].notna().all()
assert clv_analysis["first_order_promo"].notna().all()
assert clv_analysis["user_id"].is_unique

# Skenario margin (asumsi, bukan data aktual)
margin_scenarios = {
    "Margin 10%": 0.10,
    "Margin 15%": 0.15,
    "Margin 20%": 0.20
}

# Buat tabel nilai pelanggan per segmen
clv_proxy_summary = (
    clv_analysis.groupby("promo_segment")
    .agg(
        total_users=("user_id", "nunique"),
        avg_customer_value_90d=(
            "customer_value_90d", "mean"
        ),
        avg_first_promo=(
            "first_order_promo", "mean"
        ),
        repeat_rate_90d=("repeat_90d", "mean")
    )
    .reset_index()
)

clv_proxy_summary["repeat_rate_90d"] *= 100

# Dua pendekatan alternatif
for label, margin in margin_scenarios.items():

    # A: asumsi total_amount sebelum diskon
    clv_proxy_summary[f"A - {label}"] = (
        clv_proxy_summary["avg_customer_value_90d"]
        * margin
        - clv_proxy_summary["avg_first_promo"]
    )

    # B: asumsi total_amount sudah setelah diskon
    clv_proxy_summary[f"B - {label}"] = (
        clv_proxy_summary["avg_customer_value_90d"]
        * margin
    )

print("=== CLV PROXY 90 HARI - PER PELANGGAN ===")
print("A = Margin nilai transaksi - promo pertama")
print("B = Margin nilai transaksi (tanpa potongan ulang)")
print("Seluruh nilai dalam Rupiah.\n")

display(clv_proxy_summary.round(2))

# Fokus pada skenario margin 15%
comparison_15 = clv_proxy_summary[
    [
        "promo_segment",
        "total_users",
        "avg_customer_value_90d",
        "avg_first_promo",
        "repeat_rate_90d",
        "A - Margin 15%",
        "B - Margin 15%"
    ]
].sort_values("A - Margin 15%", ascending=False)

print("\n=== PERBANDINGAN MARGIN 15% ===")
display(comparison_15.round(2))
```

Skenario margin 15% digunakan sebagai ilustrasi utama, dengan dua perlakuan alternatif terhadap biaya promo. Perbandingan ini bertujuan menunjukkan sensitivitas hasil terhadap asumsi akuntansi promosi, bukan menetapkan nilai CLV aktual.
